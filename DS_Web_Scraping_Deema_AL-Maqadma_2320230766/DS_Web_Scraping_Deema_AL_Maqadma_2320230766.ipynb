```python

import requests
from bs4 import BeautifulSoup
import time
import csv
import os

# ---------------------------------------------------------
# Headers to mimic a real browser
# ---------------------------------------------------------
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0 Safari/537.36'
}

# ---------------------------------------------------------
# Dictionary of categories (اسم التصنيف : رابط التصنيف)
# ---------------------------------------------------------
categories = {
    "عطور": "https://ayatstore0.com/product-category/%d8%b9%d8%b7%d9%88%d8%b1/",
    "سكن كير": "https://ayatstore0.com/product-category/%d9%85%d9%83%d9%8a%d8%a7%d8%ac/%d8%b3%d9%83%d9%86-%d9%83%d9%8a%d8%b1/",
    "كفرات": "https://ayatstore0.com/product-category/%d8%a7%d9%83%d8%b3%d8%b3%d9%88%d8%a7%d8%b1%d8%a7%d8%aa/%d9%83%d9%81%d8%b1%d8%a7%d8%aa/",
    "فرش": "https://ayatstore0.com/product-category/%d9%85%d9%83%d9%8a%d8%a7%d8%ac/%d9%81%d8%b1%d8%b4/",
    "كلشي وكلاشي": "https://ayatstore0.com/product-category/%d9%83%d9%84%d8%b4%d9%8a-%d9%88%d9%83%d9%84%d8%a7%d8%b4%d9%8a/",
    "تراجي": "https://ayatstore0.com/product-category/%d8%a7%d9%83%d8%b3%d8%b3%d9%88%d8%a7%d8%b1%d8%a7%d8%aa/%d8%aa%d8%b1%d8%a7%d8%ac%d9%8a/",
    "جواريب": "https://ayatstore0.com/product-category/%d8%a7%d9%83%d8%b3%d8%b3%d9%88%d8%a7%d8%b1%d8%a7%d8%aa/%d8%ac%d9%88%d8%a7%d8%b1%d9%8a%d8%a8/",
    "اظافر": "https://ayatstore0.com/product-category/%d8%a7%d9%83%d8%b3%d8%b3%d9%88%d8%a7%d8%b1%d8%a7%d8%aa/%d8%a7%d8%b8%d8%a7%d9%81%d8%b1/",
    "بلاشر": "https://ayatstore0.com/product-category/%d9%85%d9%83%d9%8a%d8%a7%d8%ac/%d8%a8%d9%84%d8%b4%d8%b1/",
    "حمر": "https://ayatstore0.com/product-category/%d9%85%d9%83%d9%8a%d8%a7%d8%ac/%d8%ad%d9%85%d8%b1/",
    "دباديب": "https://ayatstore0.com/product-category/%d8%a7%d9%83%d8%b3%d8%b3%d9%88%d8%a7%d8%b1%d8%a7%d8%aa/%d8%af%d8%a8%d8%a7%d8%af%d9%8a%d8%a8/",
    "رموش": "https://ayatstore0.com/product-category/%d9%85%d9%83%d9%8a%d8%a7%d8%ac/%d8%b1%d9%85%d9%88%d8%b4/",
    "خواتم": "https://ayatstore0.com/product-category/%d8%a7%d9%83%d8%b3%d8%b3%d9%88%d8%a7%d8%b1%d8%a7%d8%aa/%d8%ae%d9%88%d8%a7%d8%aa%d9%85/",
    "اكسسوارات شعر": "https://ayatstore0.com/product-category/%d8%a7%d9%83%d8%b3%d8%b3%d9%88%d8%a7%d8%b1%d8%a7%d8%aa/%d8%a7%d9%83%d8%b3%d8%b3%d9%88%d8%a7%d8%b1%d8%a7%d8%aa-%d8%b4%d8%b9%d8%b1/",
    "كلايد": "https://ayatstore0.com/product-category/%d8%a7%d9%83%d8%b3%d8%b3%d9%88%d8%a7%d8%b1%d8%a7%d8%aa/%d9%83%d9%84%d8%a7%d9%8a%d8%af/",
    "اساور": "https://ayatstore0.com/product-category/%d8%a7%d9%83%d8%b3%d8%b3%d9%88%d8%a7%d8%b1%d8%a7%d8%aa/%d8%a7%d8%b3%d8%a7%d9%88%d8%b1/",
    "ساعات": "https://ayatstore0.com/product-category/%d8%a7%d9%83%d8%b3%d8%b3%d9%88%d8%a7%d8%b1%d8%a7%d8%aa/%d8%b3%d8%a7%d8%b9%d8%a7%d8%aa/",
    "بروشات": "https://ayatstore0.com/product-category/%d8%a7%d9%83%d8%b3%d8%b3%d9%88%d8%a7%d8%b1%d8%a7%d8%aa/%d8%a8%d8%b1%d9%88%d8%b4%d8%a7%d8%aa/",
    "مسكارة": "https://ayatstore0.com/product-category/%d9%85%d9%83%d9%8a%d8%a7%d8%ac/%d9%85%d8%b3%d9%83%d8%a7%d8%b1%d8%a9/",
    "شدو": "https://ayatstore0.com/product-category/%d9%85%d9%83%d9%8a%d8%a7%d8%ac/%d8%b4%d8%a7%d8%af%d9%88/",
    "هدايا": "https://ayatstore0.com/product-category/%d8%a7%d9%83%d8%b3%d8%b3%d9%88%d8%a7%d8%b1%d8%a7%d8%aa/%d9%87%d8%af%d8%a7%d9%8a%d8%a7/",
    "نظارات": "https://ayatstore0.com/product-category/%d8%a7%d9%83%d8%b3%d8%b3%d9%88%d8%a7%d8%b1%d8%a7%d8%aa/%d9%86%d8%b8%d8%a7%d8%b1%d8%a7%d8%aa/",
    "عروض": "https://ayatstore0.com/product-category/%d8%b9%d8%b1%d9%88%d8%b6/",
    "علب": "https://ayatstore0.com/product-category/%d8%b9%d9%84%d8%a8/",
    "رجالي": "https://ayatstore0.com/product-category/%d8%b1%d8%ac%d8%a7%d9%84%d9%8a/"
}

# ---------------------------------------------------------
# Create CSV file
# ---------------------------------------------------------
filename = 'ayatstore_products.csv'
filecsv = open(filename, 'w', encoding='utf8', newline='')
csv_columns = ['category', 'name', 'price', 'old_price', 'discount', 'product_url', 'image_url', 'description']
writer = csv.DictWriter(filecsv, fieldnames=csv_columns)
writer.writeheader()

row_count = 0

# ---------------------------------------------------------
# Loop through all categories
# ---------------------------------------------------------
for cat_name, cat_url in categories.items():
    print(f"\n=== Category: {cat_name} ===")

    # أول صفحة: نجيب عدد الصفحات من الـ pagination
    r = requests.get(cat_url, headers=headers)
    soup = BeautifulSoup(r.content, "html.parser")

    pages = soup.find_all("a", class_="page-numbers")
    if pages:
        max_page = int(pages[-2].get_text())
    else:
        max_page = 1

    print(f"Total pages in {cat_name}: {max_page}")

    # ---------------------------------------------------------
    # Loop through all pages in this category
    # ---------------------------------------------------------
    for page in range(1, max_page+1):
        print(f"\n--- Scraping {cat_name} Page {page} ---")
        r = requests.get(cat_url + f"page/{page}/", headers=headers)
        soup = BeautifulSoup(r.content, "html.parser")

        products = soup.find_all("li", class_="product")

        if not products:
            print("No products found on this page")
            continue

        # Loop through all products in this page
        for p in products:
            try:
                name = p.find("h2", class_="woocommerce-loop-product__title").get_text(strip=True)
            except:
                name = "not available"

            try:
                price = p.find("span", class_="woocommerce-Price-amount").get_text(strip=True)
            except:
                price = "not available"

            try:
                old_price = p.find("del").get_text(strip=True)
            except:
                old_price = "not available"

            try:
                discount = p.find("span", class_="onsale").get_text(strip=True)
            except:
                discount = "not available"

            try:
                product_url = p.find("a")["href"]
            except:
                product_url = "not available"

            try:
                image_url = p.find("img")["src"]
            except:
                image_url = "not available"

            try:
                description = p.find("div", class_="woocommerce-product-details__short-description").get_text(strip=True)
            except:
                description = "not available"

            # Write row to CSV
            writer.writerow({
                'category': cat_name,
                'name': name,
                'price': price,
                'old_price': old_price,
                'discount': discount,
                'product_url': product_url,
                'image_url': image_url,
                'description': description
            })

            row_count += 1

        # Sleep 3 seconds between pages
        time.sleep(3)

    # Sleep 5 seconds between categories
    time.sleep(5)

# ---------------------------------------------------------
# Close CSV file
# ---------------------------------------------------------
filecsv.close()

# ---------------------------------------------------------
# Print summary
# ---------------------------------------------------------
print("\nDone! Data saved successfully.")
print("Total rows saved:", row_count)
print("File saved as:", filename)
print("Files in current directory:", os.listdir())
```
