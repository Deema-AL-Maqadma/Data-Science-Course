```python

# Read raw CSV file as text
import pandas as pd

raw = pd.read_csv('/content/drive/MyDrive/ayatstore_products.csv', header=None)

# Split the first column by comma
data = raw[0].str.split(',', expand=True)

# Check number of columns
print("Number of columns:", data.shape[1])
```

```python

# Rename columns correctly (7 columns)
data.columns = ['category','name','price','old_price','discount','product_url','image_url']

# Show first 10 rows
data.head(10)
```

```python

# Add description column with default value
data['description'] = 'not available'

# Show first 10 rows again
data.head(10)
```

```python

# Visualize raw price values before cleaning
import seaborn as sns
import matplotlib.pyplot as plt

# Histogram for raw price values (still strings)
sns.histplot(data['price'])
plt.show()

# Boxplot for raw price values
sns.boxplot(x=data['price'])
plt.show()
```

```python

# Extract numeric values from price
data['price'] = data['price'].astype(str).str.extract(r'(\d+\.?\d*)')
data['price'] = pd.to_numeric(data['price'], errors='coerce')

# Extract numeric values from old_price
data['old_price'] = data['old_price'].astype(str).str.extract(r'(\d+\.?\d*)')
data['old_price'] = pd.to_numeric(data['old_price'], errors='coerce')

# Show first 10 rows after cleaning
data.head(10)
```

```python

# Drop the first row (header row)
data = data.drop(index=0).reset_index(drop=True)

# Fill missing values
data['discount'] = data['discount'].fillna('not available')
data['image_url'] = data['image_url'].fillna('not available')
data['description'] = data['description'].fillna('not available')

# Drop rows with missing name or price
data = data.dropna(subset=['name','price'])

# Remove duplicates
data = data.drop_duplicates()

# Show dataset info and first 10 rows
print(data.info())
data.head(10)
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

# Histogram for cleaned price values
sns.histplot(data['price'], bins=20)
plt.show()

# Boxplot for cleaned price values
sns.boxplot(x=data['price'])
plt.show()
```

```python

# Normalize price column using MinMaxScaler
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
data['price_norm'] = scaler.fit_transform(data[['price']])

# Show first 10 rows after normalization
data.head(10)
```

```python

# Visualize normalized price values
import seaborn as sns
import matplotlib.pyplot as plt

# Histogram for normalized price values
sns.histplot(data['price_norm'], bins=20)
plt.show()

# Boxplot for normalized price values
sns.boxplot(x=data['price_norm'])
plt.show()
```

```python
# Save final cleaned and normalized dataset
data.to_csv('/content/drive/MyDrive/ayatstore_products_final.csv', index=False)
```
